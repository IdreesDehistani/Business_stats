## working notebook for standardizing the columns.

```python
#Medeea
import pandas as pd
import numpy as np
```

```python
#Medeea
df = pd.read_excel('Responses.xlsx')
df.shape
```

```python
#Medeea; the column we're cleaning
col2 = "Which description fits your parents' country of origin the best?"
```

```python
#Medeea;check the raw values
df[col2].value_counts(dropna=False)
```

```python
#Medeea;strip whitespace and standardize case
df[col2].astype(str).str.strip()
```

```python
#Medeea; This dictionary maps every known variant to one clean label.
mapping = {
    'both parents are from the eu': 'Both parents are from the EU',
    'both parents are from eu':     'Both parents are from the EU',

    'both parents are from outside the eu': 'Both parents are from outside the EU',

    'both parents are born and brought up in the netherlands': 'Both parents are born and brought up in the Netherlands',

    'both parents are born and brought up in germany':     'Both parents are born and brought up in Germany',
    'both parents are born and brought up in the germany': 'Both parents are born and brought up in Germany',

    'both parents are born and brought up in spain':     'Both parents are born and brought up in Spain',
    'both parents are born and brought up in the spain': 'Both parents are born and brought up in Spain',

    'both parents are born and brought up in greece':     'Both parents are born and brought up in Greece',
    'both parents are born and brought up in the greece': 'Both parents are born and brought up in Greece',
    'both parents are from greece':                       'Both parents are born and brought up in Greece',

    'one parent is from the eu and one parent is outside of the eu': 'One parent is from the EU and one parent is outside the EU',
    'one parent is from the eu and one parent is outside the eu':    'One parent is from the EU and one parent is outside the EU',

    'one parent is from greece, and one parent is outside of greece': 'One parent is from Greece and one parent is outside Greece',
}

normalized = df[col2].str.strip().str.lower()   # NaN stays NaN
df[col2] = normalized.map(mapping).fillna(df[col2])

df[col2].value_counts(dropna=False)
```

```python
#Medeea, the columns we want to convert to numbers(Likert scale)
df.columns = df.columns.str.replace(r"\s+", " ", regex=True).str.strip()
agree_cols = [
    'Scrolling feels like a habit rather than a conscious choice',
    'I feel I am in control of my feed',
    'I can influence the type of content I get on my feed',
    'The algorithm decides what I watch',
    'I think social media is fair2',
    'I trust the algorithm to recommend me content fairly',
    'I notice a lot of agreement with the content in the comments',
    'I tend to comment on content I disagree with',
    'I often see new ideas and topics on my feed',
    'When I see a new topic on my feed, I tend to scroll away',
    'My feed does not show information that others could be seeing',
    'I feel like my content gets the same number of views as content from others',
    'I feel like my content gets the same number of views as content from others with a different background (e.g., race, gender, age) than mine',
    'I think content from certain groups of people is prioritized over other groups',
    'I am informed that my feed is shaped by personal information like my age, gender or location - transparency',
    'I sometimes spend more time scrolling than I had planned'
]

# For every column in the batch: strip spaces, lowercase, translate to a number.
df[agree_cols] = df[agree_cols].apply(lambda s: s.str.strip().str.lower().map({
    'strongly disagree': 1,
    'disagree':          2,
    'neutral':           3,
    'agree':             4,
    'strongly agree':    5,
}))
```

```python
#Medeea, the columns we want to convert to numbers, but this time it's a frequency scale
freq_cols = [
    "I end up watching videos I didn't intend to watch",
    'When I see a post and open the comments, the top comments reflect what I am already thinking about',
    'I scroll through the comments to see if someone has mentioned what I am thinking about',
    'I scroll through recommendations'
]

df[freq_cols] = df[freq_cols].apply(lambda s: s.str.strip().str.lower().map({
    'never':     1,
    'rarely':    2,
    'sometimes': 3,
    'often':     4,
    'always':    5,
}))
```

```python
#Medeea; revert likert scale so that higher numbers indicate more agreement
df['The algorithm decides what I watch'] = 6 - df['The algorithm decides what I watch']
```

```python
#Medeea; continuation of the previous cell, revert likert scale so that higher numbers indicate more agreement
df['I feel like my content gets the same number of views as content from others'] = 6 - df['I feel like my content gets the same number of views as content from others']
df['I feel like my content gets the same number of views as content from others with a different background (e.g., race, gender, age) than mine'] = 6 - df['I feel like my content gets the same number of views as content from others with a different background (e.g., race, gender, age) than mine']
df["I end up watching videos I didn't intend to watch"] = 6 - df["I end up watching videos I didn't intend to watch"]
df['Scrolling feels like a habit rather than a conscious choice'] = 6 - df['Scrolling feels like a habit rather than a conscious choice']
df['I sometimes spend more time scrolling than I had planned'] = 6 - df['I sometimes spend more time scrolling than I had planned']
```

```python
#Medeea; check the values of the cleaned columns
for col in agree_cols + freq_cols:
    print(df[col].value_counts(dropna=False).sort_index(), '\n')
```

```python
#Medeea;One-hot encode the parents' origin column.
#prefix='origin' name prefix for the new columns, e.g. "origin_Both parents are from the EU"
df = pd.get_dummies(df, columns=[col2], prefix='origin', dtype=int)

# Show the new columns
print([c for c in df.columns if c.startswith('origin_')])
```

```python
#Medeea; change columns name to short versions
short_names_REVP_BC_RAD_PF = {
    # REVP: reinforced existing views and preferences
    "When I see a post and open the comments, the top comments reflect what I am already thinking about": "REVP1",
    "I scroll through the comments to see if someone has mentioned what I am thinking about": "REVP2",
    "I notice a lot of agreement with the content in the comments": "REVP3",
    "I tend to comment on content I disagree with": "REVP4",
    "I often see new ideas and topics on my feed": "REVP5",
    "When I see a new topic on my feed, I tend to scroll away": "REVP6",

    # BC: biased content
    "My feed does not show information that others could be seeing": "BC1",
    "I feel like my content gets the same number of views as content from others": "BC2",
    "I feel like my content gets the same number of views as content from others with a different background (e.g., race, gender, age) than mine": "BC3",
    "I think content from certain groups of people is prioritized over other groups": "BC4",
    "I am informed that my feed is shaped by personal information like my age, gender or location - transparency": "BC5",

    # RAD: rational and autonomous decision-making
    "I feel I am in control of my feed": "RAD1",
    "I can influence the type of content I get on my feed": "RAD2",
    "The algorithm decides what I watch": "RAD3",
    "I scroll through recommendations": "RAD4",
    "I end up watching videos I didn't intend to watch": "RAD5",
    "Scrolling feels like a habit rather than a conscious choice": "RAD6",
    "I sometimes spend more time scrolling than I had planned": "RAD7",

    # PF: perceived fairness
    "I trust the algorithm to recommend me content fairly": "PF1",
    "I think social media is fair2": "PF2",
}

# Check: which names are NOT found in the file? This should print []
print([name for name in short_names_REVP_BC_RAD_PF.values() if name not in df.columns])

df = df.rename(columns=short_names_REVP_BC_RAD_PF)
```

```python
#Medeea;save to a new file
df.to_excel('Responses_cleaned_finalMedeea.xlsx', index=False)
print('Saved -> Responses_cleaned_finalMedeea.xlsx')
```
