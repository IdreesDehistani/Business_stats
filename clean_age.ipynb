```python
import pandas as pd

df = pd.read_excel('Responses(1).xlsx')
```

```python
# Roland
# column we're cleaning
col = 'What age group are you in?'
```

```python
# Roland
# keeping only age groups 18-24 and 25-34

keep = df[col].isin(['18-24', '25-34'])
df = df[keep]

print(df[col].value_counts(dropna=False)) # listing age goups with counts and checking for NaNs
print(df.shape)
```

```python
# Roland
# saving to the cleaned file
df.to_excel('Responses(1)_cleaned.xlsx', index=False)
print('Saved -> Responses(1)_cleaned.xlsx')
```
