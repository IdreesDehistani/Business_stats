## SEM for democracy data
Bollen in his 1989 book on structural equation modeling used the following model in his analysis of these data: each latent variable is measured by three or four indicators, industrialization is measured in 1960, and democracy is measured at two timepoints (1960 and 1965).

y1
Expert ratings of the freedom of the press in 1960

y2
The freedom of political opposition in 1960

y3
The fairness of elections in 1960

y4
The effectiveness of the elected legislature in 1960

y5
Expert ratings of the freedom of the press in 1965

y6
The freedom of political opposition in 1965

y7
The fairness of elections in 1965

y8
The effectiveness of the elected legislature in 1965

x1
The gross national product (GNP) per capita in 1960

x2
The inanimate energy consumption per capita in 1960

x3
The percentage of the labor force in industry in 1960

![image info](sem-democracy.png)

```python
# Describing the model
desc = '''
    # measurement model
    ind60 =~ x1 + x2 + x3
    dem60 =~ y1 + y2 + y3 + y4
    dem65 =~ y5 + y6 + y7 + y8
    # regressions
    dem60 ~ ind60
    dem65 ~ ind60 + dem60
    # residual correlations
    y1 ~~ y5
    y2 ~~ y4 + y6
    y3 ~~ y7
    y4 ~~ y8
    y6 ~~ y8
    '''
```

```python
# reading the data. The 'political democracy data' is provided in the semopy package itself as an example.
data = semopy.examples.political_democracy.get_data()
data.head()
```

```python
#Fitting the model to data
model = semopy.Model(desc)
results = model.fit(data)
print(results)
```

```python
# inspecting the estimated model parameters
ins = model.inspect()
print(ins)
```

```python
#Model fit indices
stats = semopy.calc_stats(model)
print(stats.T)
```

```python
# Saving the SEM path diagram with estimated model parameters
# Make sure you have the 'graphviz' package installed
# Use the command in anaconda terminal: conda install python-graphviz
# There might be some issue with the python-graphviz package. But don't worry, the only sad part is, if this package is not working you will
# not get nice visualization, but all other computations will still be working.
g = semopy.semplot(model, "semopy-fit.png")
```

```python
#Semopy provides report function that takes a fitted model and generates a folder contaning a HTML file. 
# Saving the folder with name "Political Democracy"
semopy.report(model, "Political Democracy")
```
