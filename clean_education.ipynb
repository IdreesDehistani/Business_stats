```python
import pandas as pd

df = pd.read_excel('Responses(1).xlsx')
```

```python
# Roland
# column we're cleaning
col = "What's your highest/current level of education?"
```

```python
# Roland

# removing hidden spaces behind "doctorate or equivalent"
df[col] = df[col].str.strip()

# renaming answers into four groups
df[col] = df[col].replace({
    'No formal education': 'Secondary or less',
    'Secondary/high school': 'Secondary or less',
    'College/vocational training': 'Vocational',
    "Bachelor's degree": 'Bachelor',
    "Master's degree": 'Master or higher',
    'Doctorate or equivalent': 'Master or higher',
})

print(df[col].value_counts(dropna=False))
```
