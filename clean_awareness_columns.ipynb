```python
import pandas as pd

df = pd.read_excel('Responses(1).xlsx')
```

```python
# Roland
# removing hidden spaces from the column names
df.columns = df.columns.str.strip()

# the columns we're converting
likert_cols = [
    'I am informed that my feed is shaped by personal information like my age, gender or location - transparency',
    'I feel like the algorithm shapes what political topics I get to see',
    'The algorithm decides what I watch',
    'I believe that the algorithm hides certain social or political views',
    'I think content from certain groups of people is prioritized over other groups',
]

# words to numbers
likert_map = {
    'Strongly disagree': 1,
    'Disagree': 2,
    'Neutral': 3,
    'Agree': 4,
    'Strongly agree': 5,
}

# applying to each column
for col in likert_cols:
    df[col] = df[col].replace(likert_map)


for col in likert_cols:
    print(df[col].value_counts(dropna=False))
```
